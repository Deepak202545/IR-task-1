```python

# ============================================================
# TF-IDF RESEARCH LAB - COMPLETE PROJECT
# 50 DOCUMENTS | GOOGLE COLAB
# ============================================================

import pandas as pd
import numpy as np
import re
import math
import matplotlib.pyplot as plt
import seaborn as sns

from collections import Counter
from IPython.display import display, HTML

# ============================================================
# 1. DATASET CREATION - 50 DOCUMENTS
# ============================================================

documents = [
    # TECHNOLOGY
    ("D001","Technology","Machine learning enables computers to learn patterns from data and make intelligent predictions."),
    ("D002","Technology","Artificial intelligence helps machines perform tasks that normally require human intelligence."),
    ("D003","Technology","Cloud computing provides scalable storage and computing resources through internet services."),
    ("D004","Technology","Cybersecurity protects computer systems networks and digital information from cyber attacks."),
    ("D005","Technology","Data science combines statistics programming and machine learning to analyze information."),
    ("D006","Technology","Deep learning uses artificial neural networks to recognize complex patterns in data."),
    ("D007","Technology","Software development involves designing coding testing and maintaining computer applications."),
    ("D008","Technology","Computer networks connect multiple devices and enable communication and information sharing."),
    ("D009","Technology","Natural language processing allows computers to understand and process human languages."),
    ("D010","Technology","Database management systems organize store retrieve and manage structured information."),

    # EDUCATION
    ("D011","Education","Online education provides students with flexible learning opportunities through digital platforms."),
    ("D012","Education","Students improve academic performance through regular practice and effective study methods."),
    ("D013","Education","Educational technology supports interactive learning and digital classroom environments."),
    ("D014","Education","Universities provide academic programs research facilities libraries and laboratory resources."),
    ("D015","Education","Teachers develop lesson plans to improve student understanding and classroom participation."),
    ("D016","Education","Distance learning allows students to access educational resources from different locations."),
    ("D017","Education","Examinations evaluate student knowledge understanding and academic performance."),
    ("D018","Education","Digital libraries provide access to electronic books journals and research papers."),
    ("D019","Education","Project based learning develops practical technical and problem solving skills."),
    ("D020","Education","Artificial intelligence can support personalized learning and educational assessment."),

    # SPORTS
    ("D021","Sports","Cricket requires teamwork batting bowling fielding and strategic decision making."),
    ("D022","Sports","Football players improve performance through physical training and tactical preparation."),
    ("D023","Sports","Basketball develops coordination speed shooting accuracy and teamwork."),
    ("D024","Sports","Athletes require proper nutrition exercise and recovery to maintain physical fitness."),
    ("D025","Sports","Tennis players practice serving movement and coordination to improve match performance."),
    ("D026","Sports","Sports analytics uses statistical information to evaluate player performance."),
    ("D027","Sports","Swimming improves cardiovascular fitness endurance and muscular strength."),
    ("D028","Sports","Coaches develop training programs to improve athletic skills and team performance."),
    ("D029","Sports","Olympic competitions bring athletes together to represent their countries."),
    ("D030","Sports","Regular physical activity improves health fitness and mental wellbeing."),

    # HEALTH
    ("D031","Health","Regular exercise improves cardiovascular health and supports physical fitness."),
    ("D032","Health","Balanced nutrition provides essential vitamins minerals proteins and carbohydrates."),
    ("D033","Health","Doctors use medical tests to diagnose diseases and monitor patient health."),
    ("D034","Health","Hospitals provide emergency services medical treatment and patient care."),
    ("D035","Health","Adequate sleep supports memory concentration and physical recovery."),
    ("D036","Health","Preventive healthcare includes regular medical checkups and healthy lifestyle habits."),
    ("D037","Health","Mental health is important for emotional stability and overall wellbeing."),
    ("D038","Health","Medical technology supports accurate diagnosis and effective patient monitoring."),
    ("D039","Health","Healthy eating habits reduce the risk of several chronic diseases."),
    ("D040","Health","Public health programs promote disease prevention and community healthcare."),

    # ENVIRONMENT
    ("D041","Environment","Renewable energy sources reduce dependence on fossil fuels and environmental pollution."),
    ("D042","Environment","Solar energy converts sunlight into electricity using photovoltaic technology."),
    ("D043","Environment","Forests support biodiversity wildlife habitats and environmental balance."),
    ("D044","Environment","Climate change affects weather patterns agriculture and natural ecosystems."),
    ("D045","Environment","Water conservation protects freshwater resources for future generations."),
    ("D046","Environment","Recycling reduces waste and supports sustainable resource management."),
    ("D047","Environment","Air pollution affects human health animals and environmental quality."),
    ("D048","Environment","Electric vehicles help reduce transportation emissions and fuel consumption."),
    ("D049","Environment","Sustainable development balances economic growth with environmental protection."),
    ("D050","Environment","Wildlife conservation protects endangered species and natural habitats.")
]

df = pd.DataFrame(
    documents,
    columns=["Document_ID", "Category", "Text"]
)

print("="*65)
print("TF-IDF RESEARCH LAB")
print("="*65)
print("Total Documents:", len(df))
print("Total Categories:", df["Category"].nunique())
print("\nCategory Distribution:")
print(df["Category"].value_counts())

# ============================================================
# 2. TEXT PREPROCESSING
# ============================================================

stop_words = {
    "is","the","a","an","and","or","to","of","in","on",
    "for","with","from","by","into","through","are","can",
    "be","that","this","it","as","their","its","these",
    "those","has","have","which","at"
}

def preprocess(text):
    text = text.lower()
    text = re.sub(r'[^a-z\s]', '', text)
    tokens = text.split()
    tokens = [word for word in tokens if word not in stop_words]
    return tokens

df["Processed_Text"] = df["Text"].apply(preprocess)

print("\nPREPROCESSING COMPLETED")
display(df.head())

# ============================================================
# 3. VOCABULARY GENERATION
# ============================================================

vocabulary = sorted(set(
    word
    for tokens in df["Processed_Text"]
    for word in tokens
))

N = len(df)

print("\nTotal Documents:", N)
print("Total Unique Vocabulary:", len(vocabulary))

# ============================================================
# 4. TERM FREQUENCY - TF
# ============================================================

def calculate_tf(tokens):
    counts = Counter(tokens)
    total = len(tokens)

    if total == 0:
        return {}

    return {
        word: count / total
        for word, count in counts.items()
    }

tf_results = {}

for _, row in df.iterrows():
    tf_results[row["Document_ID"]] = calculate_tf(
        row["Processed_Text"]
    )

tf_matrix = pd.DataFrame(tf_results).fillna(0)
tf_matrix = tf_matrix.reindex(vocabulary).fillna(0)

print("\nTERM FREQUENCY MATRIX")
display(tf_matrix.head(15))

# ============================================================
# 5. DOCUMENT FREQUENCY - DF
# ============================================================

df_frequency = {}

for word in vocabulary:
    df_frequency[word] = sum(
        1 for tokens in df["Processed_Text"]
        if word in set(tokens)
    )

df_table = pd.DataFrame({
    "Term": list(df_frequency.keys()),
    "Document_Frequency": list(df_frequency.values())
})

df_table = df_table.sort_values(
    "Document_Frequency",
    ascending=False
)

print("\nDOCUMENT FREQUENCY TABLE")
display(df_table.head(20))

# ============================================================
# 6. INVERSE DOCUMENT FREQUENCY - IDF
# ============================================================

idf_values = {
    word: math.log(N / df_frequency[word])
    for word in vocabulary
    if df_frequency[word] > 0
}

idf_table = pd.DataFrame({
    "Term": vocabulary,
    "DF": [df_frequency[word] for word in vocabulary],
    "IDF": [idf_values[word] for word in vocabulary]
})

idf_table = idf_table.sort_values(
    "IDF",
    ascending=False
)

print("\nINVERSE DOCUMENT FREQUENCY TABLE")
display(idf_table.head(20))

# ============================================================
# 7. TF-IDF CALCULATION
# ============================================================

tfidf_matrix = tf_matrix.copy()

for word in vocabulary:
    tfidf_matrix.loc[word] = (
        tf_matrix.loc[word] * idf_values[word]
    )

print("\nTF-IDF MATRIX")
display(tfidf_matrix.head(20))

# ============================================================
# 8. TOP KEYWORD EXTRACTION
# ============================================================

def extract_keywords(document_id, top_n=5):
    scores = tfidf_matrix[document_id].sort_values(
        ascending=False
    )

    return scores[scores > 0].head(top_n)

print("\nTOP KEYWORDS FOR SAMPLE DOCUMENTS")

for doc_id in ["D001","D011","D021","D031","D041"]:
    print("\n", doc_id)
    display(extract_keywords(doc_id))

# ============================================================
# 9. MANUAL COSINE SIMILARITY
# ============================================================

def manual_cosine_similarity(vector_a, vector_b):

    dot_product = np.dot(vector_a, vector_b)

    magnitude_a = np.linalg.norm(vector_a)
    magnitude_b = np.linalg.norm(vector_b)

    if magnitude_a == 0 or magnitude_b == 0:
        return 0.0

    return dot_product / (magnitude_a * magnitude_b)

similarity_matrix = np.zeros((N, N))

for i in range(N):
    for j in range(N):

        vector_a = tfidf_matrix.iloc[:, i].values
        vector_b = tfidf_matrix.iloc[:, j].values

        similarity_matrix[i, j] = manual_cosine_similarity(
            vector_a, vector_b
        )

similarity_df = pd.DataFrame(
    similarity_matrix,
    index=df["Document_ID"],
    columns=df["Document_ID"]
)

print("\nDOCUMENT SIMILARITY MATRIX")
display(similarity_df.iloc[:10, :10])

# ============================================================
# 10. FIND SIMILAR DOCUMENTS
# ============================================================

def find_similar_documents(document_id, top_n=5):

    scores = similarity_df[document_id].drop(document_id)

    top_results = scores.sort_values(
        ascending=False
    ).head(top_n)

    result = pd.DataFrame({
        "Document_ID": top_results.index,
        "Similarity": top_results.values
    })

    result["Similarity_Percentage"] = (
        result["Similarity"] * 100
    ).round(2)

    return result

print("\nSIMILAR DOCUMENTS FOR D001")
display(find_similar_documents("D001"))

# ============================================================
# 11. TF-IDF SEARCH ENGINE
# ============================================================

def search_documents(query, top_n=10):

    query_tokens = preprocess(query)

    query_tf = calculate_tf(query_tokens)

    query_vector = np.array([
        query_tf.get(word, 0) * idf_values[word]
        for word in vocabulary
    ])

    scores = []

    for doc_id in tfidf_matrix.columns:

        document_vector = tfidf_matrix[doc_id].values

        score = manual_cosine_similarity(
            query_vector,
            document_vector
        )

        scores.append((doc_id, score))

    scores.sort(key=lambda x: x[1], reverse=True)

    results = []

    for doc_id, score in scores[:top_n]:

        document = df[
            df["Document_ID"] == doc_id
        ].iloc[0]

        results.append({
            "Document_ID": doc_id,
            "Category": document["Category"],
            "Text": document["Text"],
            "Similarity_Percentage": round(score * 100, 2)
        })

    return pd.DataFrame(results)

print("\nSEARCH RESULTS")
display(
    search_documents(
        "machine learning artificial intelligence",
        10
    )
)

# ============================================================
# 12. COMPLETE DOCUMENT ANALYSIS
# ============================================================

def analyze_document(document_id):

    document = df[
        df["Document_ID"] == document_id
    ].iloc[0]

    print("\n" + "="*65)
    print("DOCUMENT ANALYSIS:", document_id)
    print("="*65)

    print("Category:", document["Category"])
    print("\nOriginal Text:")
    print(document["Text"])

    print("\nProcessed Tokens:")
    print(document["Processed_Text"])

    print("\nTop 10 Keywords:")
    display(
        tfidf_matrix[document_id]
        .sort_values(ascending=False)
        .head(10)
    )

    print("\nSimilar Documents:")
    display(find_similar_documents(document_id))

analyze_document("D001")

# ============================================================
# 13. TOP TF-IDF TERMS GRAPH
# ============================================================

selected_document = "D001"

top_terms = tfidf_matrix[
    selected_document
].sort_values(ascending=False).head(10)

plt.figure(figsize=(10, 5))
sns.barplot(x=top_terms.values, y=top_terms.index)
plt.title("Top 10 TF-IDF Terms - " + selected_document)
plt.xlabel("TF-IDF Score")
plt.ylabel("Terms")
plt.tight_layout()
plt.show()

# ============================================================
# 14. CATEGORY DISTRIBUTION GRAPH
# ============================================================

plt.figure(figsize=(8, 6))

df["Category"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Distribution of 50 Documents")
plt.ylabel("")
plt.show()

# ============================================================
# 15. TF-IDF HEATMAP
# ============================================================

plt.figure(figsize=(16, 9))

sns.heatmap(
    tfidf_matrix.iloc[:30, :20],
    cmap="YlGnBu",
    linewidths=0.2
)

plt.title("TF-IDF Matrix Heatmap")
plt.xlabel("Documents")
plt.ylabel("Terms")
plt.tight_layout()
plt.show()

# ============================================================
# 16. IDF DISTRIBUTION GRAPH
# ============================================================

plt.figure(figsize=(10, 5))

sns.histplot(
    list(idf_values.values()),
    bins=15,
    kde=True
)

plt.title("IDF Score Distribution")
plt.xlabel("IDF Value")
plt.ylabel("Number of Terms")
plt.tight_layout()
plt.show()

# ============================================================
# 17. EXPORT RESULTS
# ============================================================

df.to_csv("tfidf_50_documents.csv", index=False)
tf_matrix.to_csv("tf_matrix.csv")
idf_table.to_csv("idf_table.csv", index=False)
df_table.to_csv("df_table.csv", index=False)
tfidf_matrix.to_csv("tfidf_matrix.csv")
similarity_df.to_csv("similarity_matrix.csv")

print("\n" + "="*65)
print("PROJECT EXECUTION COMPLETED SUCCESSFULLY")
print("="*65)

print("Total Documents:", len(df))
print("Total Categories:", df["Category"].nunique())
print("Total Vocabulary:", len(vocabulary))
print("TF Calculation: Completed")
print("DF Calculation: Completed")
print("IDF Calculation: Completed")
print("TF-IDF Calculation: Completed")
print("Cosine Similarity: Completed")
print("Search Engine: Completed")
print("Visualizations: Completed")
print("Export: Completed")

print("\nGenerated Files:")
print("1. tfidf_50_documents.csv")
print("2. tf_matrix.csv")
print("3. df_table.csv")
print("4. idf_table.csv")
print("5. tfidf_matrix.csv")
print("6. similarity_matrix.csv")
```
